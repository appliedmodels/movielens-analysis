```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType, TimestampType
from pyspark.sql.functions import date_format

movielens_base = "/Volumes/workspace/default/movielens/"
```

```python
# genome-scores: relevance of each tag for each movie; key = (movieId, tagId)
genome_scores_schema = StructType([
    StructField("MOVIE_ID",   IntegerType(), nullable=False),
    StructField("TAG_ID",     IntegerType(), nullable=False),
    StructField("RELEVANCE", DoubleType(),  nullable=True),
])

# genome-tags: one row per tag; tagId = primary key
genome_tags_schema = StructType([
    StructField("TAG_ID", IntegerType(), nullable=False),
    StructField("TAG", StringType(), nullable=True),
])

# link:
link_schema = StructType([
    StructField("MOVIE_ID", IntegerType(), nullable=False),
    StructField("IMDB_ID", StringType(), nullable=True),
    StructField("TMDB_ID", IntegerType(), nullable=True)
])

# movie: 
movie_schema = StructType([
    StructField("MOVIE_ID", IntegerType(), nullable=False),
    StructField("TITLE", StringType(), nullable=True),
    StructField("GENRES", StringType(), nullable=True)
])

# rating: 
rating_schema = StructType([
    StructField("USER_ID", IntegerType(), nullable=False),
    StructField("MOVIE_ID", IntegerType(), nullable=False),
    StructField("RATING", DoubleType(), nullable=True),
    StructField("RATING_TIMESTAMP", TimestampType(), nullable=True)
])

# tag:
tag_schema = StructType([
    StructField("USER_ID", IntegerType(), nullable=False),
    StructField("MOVIE_ID", IntegerType(), nullable=False),
    StructField("TAG", StringType(), nullable=True),
    StructField("TAG_TIMESTAMP", TimestampType(), nullable=True)
])

genome_scores = spark.read.csv(movielens_base + "genome_scores.csv", header=True, schema=genome_scores_schema, mode="FAILFAST")
genome_tags = spark.read.csv(movielens_base + "genome_tags.csv", header=True, schema=genome_tags_schema, mode="FAILFAST")
link = spark.read.csv(movielens_base + "link.csv", header=True, schema=link_schema, mode="FAILFAST")
movie = spark.read.csv(movielens_base + "movie.csv", header=True, schema=movie_schema, mode="FAILFAST")
rating = spark.read.csv(movielens_base + "rating.csv", header=True, schema=rating_schema, mode="FAILFAST", timestampFormat="yyyy-MM-dd HH:mm:ss")
tag = spark.read.csv(movielens_base + "tag.csv", header=True, schema=tag_schema, mode="FAILFAST", timestampFormat="yyyy-MM-dd HH:mm:ss")


display(genome_tags.limit(5))
display(genome_scores.limit(5))
display(link.limit(5))
display(movie.limit(5))
display(rating.withColumn("RATING_TIMESTAMP", date_format("RATING_TIMESTAMP", "yyyy-MM-dd HH:mm:ss")).limit(5))
display(tag.withColumn("TAG_TIMESTAMP", date_format("TAG_TIMESTAMP", "yyyy-MM-dd HH:mm:ss")).limit(5))

```
